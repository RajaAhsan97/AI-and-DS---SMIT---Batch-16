# Q1: Write a program to check if the given number is +ve, -ve or 0

```python
number_ = 5

if number_ > 0:
    print("Number is Positive")
elif number_ < 0:
    print("Number is Negative")
else:
    print("Number is Zero")
```

# Q3: Write a program to find the largest of the three numbers

```python
a = 100
b = 500
c = 1000

if a > b and a > c:
    print("a is largest")
elif b > a and b > c:
    print("b is largest")
else:
    print("c is largest")
```

# Q5: Write a program to take user’s age as input and display whether he is minor, adult or senior citizen:
a.	Minor age is less than 18    
b.	Adult age is greater than 18 and less than 60  
c.	Senior citizen age is greater than 60  


```python
age = 90

if age < 18:
    print("Minor")
elif age > 18 and age < 60:
    print("Adult")
else:
    print("Senior")
```

# Q9 Write a program that ask for an integer number and checks if it is divisible by 2, 3, or both.

```python
num = 1

if num % 2 == 0 and num % 3 == 0:
    print(f"{num} Divisible by 2 and 3")
elif num % 2 == 0:
    print(f"{num} Divisible by 2")
elif num % 3 == 0:
    print(f"{num} Divisible by 3")
else:
    print(f"{num} is not divisible bt 2 or 3")
```

# Q13. Write a program to input salary and deduct Tax and calculate gross salary    
a. if salary is greater than 2500000           ---> 15% Tax Deduction  
b. if salary is between 1500000 - 2500000      ---> 10% Tax Deduction  
c. if salary is between  500000  - 1500000     --->   5% Tax Deduction  
d. if salary is less than 500000               --->   0% Tax Deduction  


```python
salary = 3000000

if salary >= 2500000:
    tax_amount = salary * 0.15
elif salary >= 1500000 and salary < 2500000:
    tax_amount = salary * 0.10
elif salary >= 500000 and salary < 1500000:
    tax_amount = salary * 0.05
else:
    tax_amount = salary * 0

gross_salary = salary - tax_amount

print(f"Salary is Rs.{salary}")
print(f"Tax Deduction is Rs.{tax_amount}")
print(f"Net Amount is Rs.{gross_salary}")
```
