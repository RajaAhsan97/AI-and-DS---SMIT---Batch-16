# Dictionary
    - {}
    - {key:value}
    - no index
    - mutable

```python
record = {'Name': "Shameer", "Course": "AI and DS", "Batch": 16}
```

```python
record
```

```python
print(type(record))
```

```python
len(record)
```

```python
# accessing value by key (not by index)
record['Name']
```

```python
# value assignment to the key
record['Name'] = "Talha"
```

### Adding new key:value pair

```python
record
```

```python
record["Campus"] = "Hajiyani Sakeena"
record["Location"] = "New Town"
```
