```python
class Student:
    # constructor
    def __init__(self, name, age, rollno, email, gender):
        # instance attributes
        self.__name = name
        self.__age = age
        self.__rollno = rollno
        self.__email = email
        self.__gender = gender

    # instance methods
    def attend_class(self):
        print(f"{self._name} can attend class")

    def appear_exam(self):
        print(f"{self._name} bearing {self._rollno} appears in exam")

    def result(self):
        print(f"{self._name} is pass in exam")

    # getter
    def get_roll_number(self):
        return self.__rollno

    # setter
    def set_roll_number(self, new_rollno):
        self.__rollno = new_rollno

    def get_name(self):
        return self.__name

    # setter
    def set_name(self, new_name):
        blacklisted_name = "ahsan"

        if new_name in blacklisted_name:
            print("This name is not allowed")
        else:
            self.__name = new_name
```

# Inheritance

Think of inheritance like a family tree.

A child inherits traits (like eye color, height) from parents.

Similarly, in programming, a child class inherits attributes and methods from a parent class.

    - If Class A inherits Class B, then Class B is called Parent Class and Class A is called Child class
    - The child class will have access to everything of parent class

### Objective 1: Demonstration of How Child Class Inherit the Attributes and Methods from Parent CLass

```python
# Parent class
class Car():
    """A simple attempt to represent a car."""
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0
        
    def get_descriptive_name(self):
        long_name = str(self.year) + ' ' + self.make + ' ' + self.model
        return long_name.title()
    
    def read_odometer(self):
        print("This car has " + str(self.odometer_reading) + " miles on it.")
    
    def update_odometer(self, mileage):
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can't roll back an odometer!")
    
    def increment_odometer(self, miles):
        self.odometer_reading += miles
```

```python
# Child class
class ElectricCar(Car):
    pass
```

### Objective 4: Demonstration of How child Class can override the methods of Parent Class and Child class has its own methods

```python
class ElectricCar(Car):

    def __init__(self, make, model, year):
        """
        Initialize attributes of the parent class.
        Then initialize attributes specific to an electric car.
        """
        super().__init__(make, model, year)
        self.battery_size = 70

    def describe_battery(self):
        """Print a statement describing the battery size."""
        print("This car has a " + str(self.battery_size) + "-kWh battery.")

    def get_descriptive_name(self):
        long_name = 'This is Electric car: ' + str(self.year) + ' ' + self.make + ' ' + self.model
        return long_name.title()
    
```
