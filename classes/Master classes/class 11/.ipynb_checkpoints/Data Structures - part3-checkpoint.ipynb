```python
# intersection
setA = {1,2,3,4,5,6,7,8,9}
setB = {1,3,5,7,9}

setA.intersection(setB)
```

```python
setA.intersection_update(setB)
```

```python
setA
```

```python
# union
setA = {1,2,3,4,5,6,7,8,9}
setB = {1,3,5,7,9}

setA.union(setB)
```

```python
# disjoint

# agr setA ky tamam members setB m mojood nai hein ----> then True
```

```python
setA.isdisjoint(setB)
```

```python
setC = {100,200,300}
```

```python
setC.isdisjoint(setB)
```
