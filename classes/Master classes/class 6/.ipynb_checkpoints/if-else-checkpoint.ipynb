#### Exercise 1: Shopping discount calculator

Enter Puchase amount, and calculate discount amount and net purchase amount  
if customer purchase amount is greater than Rs 10000 then give the discount of 10%

```python
purchase_amount = 8000

if purchase_amount >= 10000:   # False
    # calculate 10% discount
    discount = purchase_amount * 0.10
    # calculate net amount
    net_amount = purchase_amount - discount

    print("Mubarak ho! AP ko discount mil gya")
    print(f"Discount: Rs {discount}")
    print(f"Net Purchase Amount: Rs {net_amount}")
else:
    print("AP ko DISCOUNT nai mil sakta")
```

#### Exercise 2: TAX Deduction -- PART 1

Enter Salary if Salary is greater then Rs 600000 then deduct 5% Tax and calculate Gross Salary

```python
net_salary = int(input("Enter Net Salary"))

if net_salary >= 600000:
    print("AP ka Tax Deduct hoga")

    tax_amount = net_salary * 0.05
    gross_salary = net_salary - tax_amount

    print(f"AP ka itna Rs {tax_amount} kata h")
    print(f"AP ko yeh Salary Rs {gross_salary} milygi")
else:
    print("Ap ka tax nai kata...")
```

#### Exercise 3: TAX Deduction -- PART 2

Input Salary and calculate Gross Salary after TAX deduction  
    a. 2500000            ---> 15% Tax  
	b. 1500000 - 2500000  ---> 10% Tax  
	c. 500000  - 1500000  --->  5% Tax  
	d. <500000            --->  0% Tax  

```python
salary = 3000000

if salary >= 2500000:
    tax_amount = salary * 0.15
elif salary >= 1500000:
    tax_amount = salary * 0.10
elif salary >= 500000:
    tax_amount = salary * 0.05
else:
    tax_amount = salary * 0

gross_salary = salary - tax_amount

print(f"Ap ka {tax_amount} TAX kata h")
print(f"AP ka Gross Salary {gross_salary} milygi")
```
