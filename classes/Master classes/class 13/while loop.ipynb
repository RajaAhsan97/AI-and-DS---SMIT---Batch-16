# While loop:
    - jb humein loop ki iteration pata nai hoti ky loop ko kitni baar chalana h. Bulky loop ko kisi condition py khatam ho to hum while loop istamal krty hein

```python
while True:
    print("Mein loop m hun...")
```

Jb while loop ko condition True mili h to wo continuously apni body me chalta rehta h (infinitely).
Agr while loop ko terminate krwana h to condition false honi chai h

```python
i = 0

while i < 10:
    print("Hello World")
    i+=1
```

```python
# task 1
number = 1

while number <= 30:
    print(number)
    number+=1
```

```python
# task 8
number = 10

while number > 0:
    print(number)
    number-=1
```

```python
# preparing guest list

guests = []
count = 0

while count < 5:
    guest_name = input("Enter guest name: ")

    if guest_name == "exit":
        break
    
    guests.append(guest_name)
```
