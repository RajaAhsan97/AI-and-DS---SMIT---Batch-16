# File Handling:

A file is a sequence of bytes stored on a secondary memory device to which Python
can interact with. This interaction is known as file handling.

Files are of various types:  
 Text file (.txt, .csv, .docx)  
 Binary file (.bin, .exe, .png, .mp3)  
 Spreadsheet (.xlsx)  

```python
f = open("myfile.txt", "r")

content = f.read()

print(content)
```

```python
content = f.read()

print(content)
```

```python
f.close()
```

## Context manager

```python
with open("myfile.txt", "r") as f:
    content = f.read()

    print(content)
```

```python
f.read()
```

```python
with open("myfile.txt", "r") as f:
    content = f.read(8)

print(content)
```

```python
with open("myfile.txt", "r") as f:
    content = f.readline()

print(content)
```

```python
# read all lines into a list
with open("myfile.txt", "r") as f:
    content = f.readlines()

print(content)
```

```python
with open("myfile.txt", "r") as f:
    content = f.readlines()

    for i, line in enumerate(content):
        print(f"line {i+1}: {line}")
```

```python
with open("myfile.txt", "r") as f:
    content = f.readlines()

    for i, line in enumerate(content):
        if ("class" in line) or ("object" in line):
            print(f"line {i+1}: {line}")
```

```python
# Append mode -> no read and write in append mode
with open("myfile.txt", "a") as f:
    f.write("I love Python")
```

```python
with open("myfile.txt", "r") as f:
    content = f.read()

print(content)
```

```python
# Write mode --> erase the existing content and writes from the start of file

with open("myfile.txt", "w") as f:
    f.write("I love Python")
```

```python
lines = ["I love Python", "I love Deep Learning"]
with open("myfile.txt", "w") as f:
    f.writelines(lines)
```

```python
with open("guests.txt", "w") as f:
    serial = 0
    while True:
        guest = input("Enter guest name or {q} to exit").lower()
        serial += 1
        
        if guest == "q":
            break

        f.write(f"{serial}: {guest}\n")
```

```python
with open("guests.txt", "r") as f:
    content = f.read()

    print(content)
```

```python
# add now guests --> append mode

with open("guests.txt", "a") as f:
    serial = 0
    while True:
        guest = input("Enter guest name or {q} to exit").lower()
        serial += 1
        
        if guest == "q":
            break

        f.write(f"{serial}: {guest}\n")
```

```python
with open("myfile.txt", "r+") as f:
    print(f.read())

    f.write("\nThis is a new line")

    f.seek(0)

    print(f.read())
```

```python
with open("newfile.txt", "w+") as f:
    f.write("This is line1")
    f.write("\nThis is line2")
    f.write("\nThis is line3")
    f.seek(0)
    print(f.read())
```

```python
with open("guests.txt", "a+") as f:
    f.seek(0)
    lines = f.readlines()

    serial = len(lines) 
    while True:
        guest = input("Enter guest name or {q} to exit").lower()
        serial += 1
        
        if guest == "q":
            break

        f.write(f"{serial}: {guest}\n")
```

```python
with open("guests.txt", "r") as f:
    print(f.read())
```
