if-elif-else

Input Salary and deduct tax and then calculate gross slary

1. if salary >2500000  --> 15%
2. if slaary 1500000 - 2500000 ---> 10 %
3. if slaary 500000 - 1500000  ---> 5%
4. if < 500000   ---> 0%

```python
salary= 500000

if salary > 2500000:
    tax_amount = salary * 0.15
elif salary > 1500000 and salary < 2500000:
    tax_amount = salary * 0.10
elif salary > 500000 and salary < 1500000:
    tax_amount = salary * 0.05
else:
    tax_amount = salary * 0

gross_salary = salary - tax_amount

print(f"Tax deduction {tax_amount}")
print(f"Gross Salary {gross_salary}")
```
