```python
# list
students = ["Hassan", "Ibrahim", "Ahmed", "Saad", "Husnain"]
```

```python
students[3]
```

```python
# iterating over list using Index number
for i in range(5):
    print(f"Hello {students[i]}")
```

```python
students
```

```python
# iterating over list
for i in students:
    print(f"Hello {i}")
```

## break

```python
students
```

```python
for student in students:
    if student == "Ahmed":
        break

    print(f"Hello {student}")
```

### continue

```python
for i in range(10):
    if i == 5:
        continue
    
    print(f"Hello {i}")
```
