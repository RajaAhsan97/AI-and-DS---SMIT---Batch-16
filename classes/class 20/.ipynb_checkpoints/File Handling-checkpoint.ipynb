# File handling:


# Files:
Files can be of different forms
1. Text file (.txt, .docx )
2. Binary file (.exe, .bin, .png, .jpeg, .mp3)
3. Speadsheet (.xlsx)

```python
file = open("myfile.txt", mode="r")

print(file.read())

file.close()
```

```python
file.read()
```

```python
# with keyword
with open("myfile.txt", mode="r") as file:
    # read all data from file (from start to end)
    content = file.read()

print(content)
```

```python
with open("myfile.txt", mode="r") as file:
    # read specified characters from a file
    content = file.read(10)

print(content)
```

```python
with open("myfile.txt", mode="r") as file:
    # read a line from a file
    print(file.readline())
    print(file.readline())
```

```python
with open("myfile.txt", mode="r") as file:
    # read all lines and store them in a list
    content = file.readlines()

    for i, line in enumerate(content):
        print(f"line {i+1}: {line}")

```

```python
with open("myfile.txt", mode="r") as file:
    # read all lines and store them in a list
    content = file.readlines()

    for i, line in enumerate(content):
        line = line.lower()
        if ("class" in line) or ("object" in line):
            print(f"line {i+1}: {line}")
```

```python
# write content at the end of file
with open("myfile.txt", mode="a") as file:
    file.write("\nI code in Python")
```

```python
with open("myfile.txt", mode="r") as file:
    print(file.read())
```

```python
# write mode --> erase all content from file and write from the start of file

with open("myfile.txt", mode="w") as file:
    file.write("I code in Python")
```

```python
lines = ["This is line1", "This is line2"]

with open("myfile.txt", mode="w") as file:
    # write multiple lines in a file
    file.writelines(lines)
```

```python
with open("guests.txt", mode="w") as file:
    serial = 0
    while True:
        guest = input("Enter guest name or {q} to exit: ")

        if guest == "q":
            break
        else:
            serial += 1
            file.write(f"{serial}: {guest}\n")
        
```

```python
# r+ mode:  read and write
with open("myfile.txt", mode="r+") as file:
    print(file.read())

    file.write("\nHassan ny course leave kr dia...")

    print("Now reading whole file:")
    file.seek(0)
    print(file.read())
```

```python
# appending new names in existing files 
with open("guests.txt", mode="r+") as file:
    content = file.readlines()
    serial = len(content) # 3
    while True:
        guest = input("Enter guest name or {q} to exit: ")

        if guest == "q":
            break
        else:
            serial += 1
            file.write(f"{serial}: {guest}\n")
        
```
