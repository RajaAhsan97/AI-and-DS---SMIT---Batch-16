# There is no function overloading in Python - means two functions can not be of same names

```python
def add(a,b):
    return a+b
```

```python
def add(a,b,c):
    return a+b+c
```

```python
add(1,2,3)
```

## Arbitrary number of arguments

```python
def add_nums(a):
    print(a)
```

```python
add_nums(1,3)
```

```python
def add_nums(*a):   # * receives all values in tuple
    print(a)
```

```python
add_nums(1,3,4,5,1)
```

```python
def add_nums(*a):   # * receives all values in tuple
    return sum(a)
```

```python
add_nums(1,3,4,5,1)
```

```python
def add_even_nums(*a):   # * receives all values in tuple
    result = 0
    for n in a:
        if n%2 == 0:
            result += n

    return result
```

```python
add_even_nums(1,3,4,5,1)
```

## Keyword arguments

```python
def make_my_profile(name, subject, qualification, experience, **other_info):
    profile = {}
    profile["Name"] = name
    profile["Subject"] = subject
    profile["Qualification"] = qualification
    profile["Experience"] = experience
    profile.update(other_info)

    return profile
    
```

```python
make_my_profile(name="Ahsan", subject="AI", qualification="MS", experience=20,
               city="Karachi", age=10)
```
