### Constructor / Initializer:
    object ky variables m values initialize krwata h

### if we do not define any constructor, then default constructor is initialize, which can take and give no values

```python
# object Designing

class Computer:

    # constructor
    def __init__(self, ram, processor, storage, price, model):
        # characteristics / properties / attributes / variables
        self.ram = ram
        self.processor = processor 
        self.storage = storage
        self.price = price
        self.model = model

    # behaviours
    # METHODS
    def coding(self):
        print(f"I am Coding on {self.model}...")

    def designing(self):
        print(f"I am Designing {self.model}...")

    def browsing(self):
        print(f"I am Browsing on {self.model}...")
```

```python
comp1 = Computer()
```

```python
comp1 = Computer("8GB", "corei7", "1TB", "50000", "DELL")
```

```python
comp2 = Computer("16GB", "corei9", "2TB", "90000", "HP")
```

```python
comp1
```

```python
comp1.coding()
```

```python
comp2.coding()
```

```python
Computer.browsing(comp1)
```

```python
comp1.browsing()
```

## Methods:
    functions defined inside class is called methods 

    Two types of methods:
        1. class methods:
                if no self in methods
        2. Instance methods:
                if self in methods

```python
class Students:
    # constructors
    def __init__(self, roll, name, course, gender, email):
        # attributes
        self.roll = roll
        self.name = name
        self.course = course
        self.gender = gender
        self.email = email

    # Instance methods
    def attend_class(self):
        print(f"{self.name} is attending class")

    def appear_exam(self):
        print(f"{self.name} {self.roll} is appearing in exam")

    def result(self):
        print(f"{self.name} is Pass in exam")
```

```python
s1 = Students("AP01", "Ahsan", "Python Advance", "Male", "xyz@gmail.com")
```

```python
s1
```

```python
s2 = Students("AP02", "Ehsan", "Python Advance", "Male", "XYZ@gmail.com")
```

```python
s1.attend_class()
```

```python
s2.attend_class()
```

```python
s1.name = "saad"
```

```python
s1.attend_class()
```
