```python
def circle(radius):
    """
        Compute area and circumference of a circle

        Parameters:
            - radius

        Return:
            - area
            - circumference
    """
    area = 3.14 * radius ** 2
    circumference = 2 * 3.14 * radius

    return area, circumference
```

```python
a, c = circle(5)
```

```python
print(f"Area of circke: {a}")
print(f"Circumference of circle: {c}")
```

### Scope of variables

```python
variable = 9    # Global variable

def var():
    variable = 8      # local variables
    print(variable)

var() 
print(variable)
```

```python
var()
```

```python
# function definition
def calculator(a, b, operator):
    if operator == "+":
        return a + b
    elif operator == "-":
        return a-b
    elif operator == "/":
        return a /b
    elif operator == "*":
        return a * b
```

```python
result = calculator(2,3, "*")
print(result)
```
