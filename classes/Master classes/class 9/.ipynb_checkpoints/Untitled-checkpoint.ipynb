### SLICING: Getting (copy) subset of a list

```python
# +ve          0        1         2         3        4         5         6         7        
students = ["Ahsan", "Talha", "Hammad", "Hassan", "Javaid", "Amjad", "Sheryar", "Yahya"]
# -ve         -8       -7        -6         -5       -4        -3        -2        -1 
```

```python
students[2:5]   # last index 5th is not included
```

```python
students[4:7]
```

```python
students[4:8]
```

```python
students[4:]
```

```python
students[-7:-2]
```

```python
students[-6:7]
```

```python
students[7:2]
```

```python
students[4:-4]
```

```python
students[-8:1]
```

```python
students[0:7:1]   # start : end : step
```

```python
students[0:7:2]
```

```python
students[::2]
```

```python
students[::-1]
```

```python
students[7:3:-1]
```

## LIST Functions

```python
# +ve          0        1         2         3        4         5         6         7        
students = ["Ahsan", "Talha", "Hammad", "Hassan", "Javaid", "Amjad", "Sheryar", "Yahya"]
# -ve         -8       -7        -6         -5       -4        -3        -2        -1 
```

### adding members in list
    - insert()   ---> adds member on given index
    - append()   ---> Adds a member in last
    - extend()   ---> Adds another collection one by one
    -  +         ---> Joins two list and forms a new list

```python
students.append('Saad')
```

```python
print(students)
```

```python
students.insert(4, "Hasnain")
```

```python
print(students)
```

```python
new_students = ["ali", "kashan", "mujtaba"]
```

```python
students.append(new_students)
```

```python
print(students)
```

```python
del students[-1]
```

```python
students.extend(new_students)
```

```python
print(students)
```

```python
students + [1,2,3,4,5]
```

```python
print(students)
```

### DELETE a member from list
    - del:  delete by index (from memory)
    - pop():  delete by index
    - remove(): delete by value
    - clear(): delete all elements

```python
students
```

```python
del students[3]
```

```python
students.remove("Yahya")
```

```python
students
```

```python
students.pop() # returns the deleted item
```

```python
students.pop(5)
```

```python
students.clear()
```
