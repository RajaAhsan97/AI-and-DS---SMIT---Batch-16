## try: Except, else, finally

1. Syntax error --> during programming hum es ko resolve kr lety hein
2. Logical error

#### Sab sy pehly humein code m wo lines identify krni hoti hein jahan py error any ky chances hein, phir us error ko humein handle krna perta h

```python
try:
    file_name = input("Enter file name with extension: ")
    with open(file_name) as file:
        print(f.read())
except FileNotFoundError:
    print("file is not present in directory")
```

```python
try:
    file_name = input("Enter file name with extension: ")
    lst = []
    print(lst[3])
    with open(file_name) as file:
        print(f.read())
except FileNotFoundError:
    print("file is not present in directory")
except IndexError:
    print("List index out of range")
```

```python
try:
    n1 = int(input("Enter a number: "))
    n2 = int(input("Enter a number: "))

    result = n1/n2
    print(result)

except Exception as e:
    print("Exception caught", e)
```

```python
try:
    n1 = int(input("Enter a number: "))
    n2 = int(input("Enter a number: "))

    r = n1/n2
    print(r)

except Exception as e:
    print("Exception caught", e)

else:
    print(f"Result is {r}")
```

```python
try:
    n1 = int(input("Enter a number: "))
    n2 = int(input("Enter a number: "))

    r = n1/n2
    print(r)

except Exception as e:
    print("Exception caught", e)

else:
    print(f"Result is {r}")

finally:
    print("that runs in every case.")
```

```python
try:
    n1 = int(input("Enter a number: "))
    n2 = int(input("Enter a number: "))

    r = n1/n2
    print(r)

except Exception as e:
    print("Exception caught", e)

else:
    print(f"Result is {r}")

finally:
    print("that runs in every case.")
```

### Use case of finally

Its most critical real-world use case is resource cleanup, such as closing a database connection, freeing network sockets, or releasing file handles

except block, skipping any normal close procedures. Placing the file closure in a finally block ensures the file never leaks out of memory.

```python
try:
    f = open("test.txt", "w+")
    age = input("Enter age: ")
    f.write(age)
    name = input("Enter name: ")
    f.write(name)
    weight = int(input("Enter weight"))
    f.write(weight)
except Exception as e:
    print(e)
finally:
    f.close()
```

### User defined Exceptions:

```python
age = int(input("Enter student age: "))    # 15<age<70

if (age < 15) or (age > 70):
    raise Exception("Student with age less then 15 and greater then 70 is not eligible")
```

```python
try:
    age = int(input("Enter student age: "))    # 15<age<70
    
    if (age < 15) or (age > 70):
        raise Exception("Student with age less then 15 and greater then 70 is not eligible")
except Exception as e:
    print(e)
```
