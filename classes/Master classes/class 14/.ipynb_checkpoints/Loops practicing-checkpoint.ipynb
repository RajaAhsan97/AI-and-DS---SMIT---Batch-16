# Q11

```python
even_sum = 0
odd_sum = 0

for i in range(51):
    if i % 2 == 0:
        even_sum += i
    else:
        odd_sum += i

print(f"Even sum: {even_sum}")
print(f"Odd sum: {odd_sum}")
```

```python
even_sum = 0
odd_sum = 0
i = 1

while i <= 50:
    if i %2 == 0:
        even_sum += i
    else:
        odd_sum += i

    i += 1

print(f"Even sum: {even_sum}")
print(f"Odd sum: {odd_sum}")
```

# Q12

```python
guests = []

for i in range(10):
    name = input("Enter name: ")

    if name in guests:
        print("Name already exists.")
        continue

    guests.append(name)
```

```python
guests
```

```python
guests = []

while True:
    name = input("Enter name: ")

    if name in guests:
        print("Name already exists.")
        continue

    if name == "exit":
        break

    guests.append(name)

print(guests)
```

# Q14

```python
discard_menu = ["fish", "qeema", "nihari"]
dishes = []
number_of_dishes = 5
i = 0

while True:
    dish = input("Dish ka naam btaen: ").lower()
    
    if dish in discard_menu:
        print("Ye DIsh discarded h...")
        continue

    dishes.append(dish)
    i += 1
    print(f"menu {i} select hogya")

    if i == number_of_dishes:
        break

print(dishes)
```

```python
dishes_cost = {}

for dish in dishes:
    cost = int(input(f"{dish} ki cost dalen: "))

    dishes_cost[dish] = cost
```

```python
dishes_cost
```

```python
total_price = 0

for price in dishes_cost.values():
    total_price += price

print(f"TOtal price: {total_price}")
```

```python
mehman = 300

menu_price_per_head = {}

for dish in dishes:
    dish_per_head = dishes_cost[dish] / mehman

    menu_price_per_head[dish] = dish_per_head
```

```python
menu_price_per_head
```

```python
per_head = sum(menu_price_per_head.values())
```

```python
per_head * 300
```
