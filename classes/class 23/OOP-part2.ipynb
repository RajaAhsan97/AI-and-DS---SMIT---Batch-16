# Object Oriented programming:

## What is Class?
it is blueprint/template/design of an object

```python
# Blueprint of an object
class Student():
    # constructor 
    def __init__(self, name, rollno, email, gender, age):
        # instance attributes/PROPERTIES
        self.name = name
        self.rollno = rollno
        self.email = email
        self.gender = gender
        self.age = age
        

    # instance methods
    def attend_class(self):
        print(f"{self.name} can attend class")

    def appear_exam(self):
        print(f"{self.name} {self.rollno} appears in exam")
```

```python
s1 = Student("Ahsan", "AI-098", "xyz@gmail.com", "Male", 23)
s2 = Student("Saad", "AI-100", "xyz@gmail.com", "Male", 30)
```

### What is Encapsulation?
It is a programming practice in which we bundle attributes and behaviour in a class. 
We restrict access to the attributes

### Access modifiers:

1. Public
2. Private  (__)
3. Protected (_)

```python
# Blueprint of an object
class Student():
    # constructor 
    def __init__(self, name, rollno, email, gender, age):
        # instance attributes/PROPERTIES
        # Public attributes
        self.name = name
        self.__rollno = rollno
        self.__email = email
        self.gender = gender
        self.age = age
        
    # getter and setter methods
    def get_rollno(self):
        return self.__rollno

    # setter method
    def set_rollno(self, new_rollno):
        blacklisted_rollno = ["AI-001", "AI-002"]

        if new_rollno in blacklisted_rollno:
            print("This rollno is not allowed")
        else:
            self.__rollno = new_rollno          
        
    # instance methods
    def attend_class(self):
        print(f"{self.name} can attend class")

    def appear_exam(self):
        print(f"{self.name} {self.rollno} appears in exam")
```

```python
s1 = Student("Ahsan", "AI-098", "xyz@gmail.com", "Male", 23)
```

```python
s1.name 
```

```python
s1.name = "Shameer"
```

```python
s1.name
```

```python
s1.rollno = "DS-097"
```

```python
s1.set_rollno("AI-100")
```
