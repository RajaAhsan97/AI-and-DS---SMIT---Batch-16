# Library Management System:

```python
books = {}
```

```python
# ADD Books
def add_book():
    id = input("Enter Book ID:")
    title = input("Enter Book Title:")
    author = input("Enter Author name:")
    stock = int(input("Enter total stock:"))

    status = books.get(id, None)

    if status is not None:
        print("Book already exists")
        return
    
    book = {
        "ID": id,
        "Title": title,
        "Author": author,
        "Total stock": stock,
        "Available copies": stock
    }

    books[id] = book

    print("Book Added successfully")
```

```python
# View all books
def view_books():
    print("---------------------- BOOKS -------------------------")
    for id, book in books.items():
        print("==================================================")
        print(f"Book ID: {id}")
        print(f"Title: {book["Title"]}")
        print(f"Author: {book["Author"]}")
        print(f"Total stock: {book["Total stock"]}")
        print(f"Available copies: {book["Available copies"]}")
        print("==================================================")
```

```python
# search book
def search_book():
    id = input("Enter Book ID to search: ")

    status = books.get(id, None)

    if status is None:
        print("Book does not exist")

        return

    print("==================================================")
    print(f"Book ID: {id}")
    print(f"Title: {status["Title"]}")
    print(f"Author: {status["Author"]}")
    print(f"Total stock: {status["Total stock"]}")
    print(f"Available copies: {status["Available copies"]}")
    print("==================================================")
    
```

```python
# Update Book
def update_book():
    book_id = input("Enter Book id to update stock:")

    status = books.get(book_id, None)

    if status is None:
        print("Book does not exist!!! Unable to update")

        return

    for id, book in books.items():
        if id == book_id:
            new_stock = int(input("Enter new stock: "))

            total_stock = books[id]["Total stock"] + new_stock
            available_stock = books[id]["Available copies"] + new_stock
            
            books[id].update({
                "Total stock": total_stock,
                "Available copies": available_stock
            })

            print("Stock updated successfully")

            return
```

```python
# Delete Book
def delete_book():
    book_id = input("Enter Book id to delete stock:")

    status = books.get(book_id, None)

    if status is None:
        print("Book does not exist!!! Unable to delete")

        return

    for id, book in books.items():
        if id == book_id:
            del books[id]
            print("Book Deleted successfully")

            return
```

```python
# Issue Book
def issue_book():
    book_id = input("Enter Book id to ISSUE book:")

    status = books.get(book_id, None)

    if status is None:
        print("Book does not exist!!! Unable to Issue Book")

        return
    
    for id, book in books.items():
        if id == book_id:

            if books[id]["Available copies"] > 0:
            
                books[id].update({
                    "Available copies": books[id]["Available copies"] - 1
                })
    
                print("Book ISSUED successfully")

                return
            else:
                print("Unable to ISSUE Book. No available copies")

                return
```

```python
# Return Book
def return_book():
    book_id = input("Enter Book id to RETURN book:")

    status = books.get(book_id, None)

    if status is None:
        print("Book does not exist!!! Unable to RETURN Book")

        return
    
    for id, book in books.items():
        if id == book_id:

            # check if availabel copies should not exceed total stock
            if books[id]["Available copies"] < books[id]["Total stock"]:
            
                books[id].update({
                    "Available copies": books[id]["Available copies"] + 1
                })
    
                print("Book RETURNED successfully")

                return
            else:
                print("Unable to RETURN Book. Stock FULL")

                return
```

```python
while True:
    print("===============================")
    print("Enter {1} to ADD new Book")
    print("Enter {2} to VIEW all Books")
    print("Enter {3} to SEARCH a Book")
    print("Enter {4} to UPDATE a Book")
    print("Enter {5} to DELETE a Book")
    print("Enter {6} to ISSUE a Book")
    print("Enter {7} to RETURN a Book")
    print("Enter {e} to EXIT")
    print("===============================")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_book()
    elif choice == "2":
        view_books()
    elif choice == "3":
        search_book()
    elif choice == "4":
        update_book()
    elif choice == "5":
        delete_book()
    elif choice == "6":
        issue_book()
    elif choice == "7":
        return_book()
    elif choice == "e":
        break
    else:
        print("INVALID choice")
    
```
