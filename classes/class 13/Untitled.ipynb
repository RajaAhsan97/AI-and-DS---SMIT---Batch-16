```python
for num in range(10):
    square = num ** 2
    print(f"Square of {num} is {square}")
```

```python
shopping_list = []

for i in range(5):
    item = input("Decide item to purchase: ")

    if item in shopping_list:
        print("Item list me mojood h")
        continue
        
    shopping_list.append(item)

```

```python
shopping_list
```

```python
items_prices = {}

for item in shopping_list:
    price = float(input(f"{item} ki price btao"))

    items_prices[item] = price
```

```python
items_prices
```

```python
items_prices.values()
```

```python
total_price = sum(items_prices.values())

print(f"Total price is {total_price}")
```

```python
total_price = 0

for price in items_prices.values():
    total_price = total_price + price  

print(f"Total price is {total_price}")
```

# While loop:

```python
shameer = 2

while shameer < 10:
    print(f"{shameer} Hello...")

    shameer += 2  # i = i + 1

```

```python
guests = []
count = 0

while count <= 5:
    name = input("Naam btao: ")

    guests.append(name)
    count += 1
```
