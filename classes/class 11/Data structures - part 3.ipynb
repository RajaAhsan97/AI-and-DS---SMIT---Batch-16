```python
a = {1,2,3,4,5,6,7,8,9}
b = {1,3,5,7,9,11}

# difference()
a.difference(b)
```

```python
# difference update
a.difference_update(b)
```

```python
a
```

```python
# union
a = {1,2,3,4,5,6,7,8,9}
b = {1,3,5,7,9,11}

c = a.union(b)
```

```python
c
```

```python
# intersection
a = {1,2,3,4,5,6,7,8,9}
b = {1,3,5,7,9,11}

a.intersection(b)
```

```python
# disjoint

a = {1,2,3,4,5,6,7,8,9}
b = {100,200,300,1}

a.isdisjoint(b)
```
