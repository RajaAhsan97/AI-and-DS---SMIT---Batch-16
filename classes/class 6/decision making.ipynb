# Exercise 1: Shopping Discount calculator

10000  ---> 10%

```python
purchase_amount = 50000

if purchase_amount >= 10000:  # True
    # discount calculation

    # 50000 * 10/100
    discount = purchase_amount * 0.10
    net_amount = purchase_amount - discount

    print("Mubarak ho AP ko discount mil gya")
    print(f"Discount: {discount}")
    print(f"Total Amount: {net_amount}")
else:
    print("AP ko discount nai mila")
```
