# Types of Inheritance:
1. Single level  --> A child class has a single parent
2. Multi level
3. Multiple inheritance

```python
# Parent class
class Car():
    """A simple attempt to represent a car."""
    # constructor
    def __init__(self, make, model, year):
        # instance attributes
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0

    # instance methods
    def get_descriptive_name(self):
        long_name = str(self.year) + ' ' + self.make + ' ' + self.model
        return long_name.title()
    
    def read_odometer(self):
        print("This car has " + str(self.odometer_reading) + " miles on it.")
    
    def update_odometer(self, mileage):
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can't roll back an odometer!")
    
    def increment_odometer(self, miles):
        self.odometer_reading += miles
```

```python
# child class
class ElectricCar(Car):
    # constructor
    def __init__(self, make, model, year, battery_size):
        """
        Initialize attributes of the parent class.
        Then initialize attributes specific to an electric car.
        """
        super().__init__(make, model, year)
        self.battery_size = battery_size

    def describe_battery(self):
        """Print a statement describing the battery size."""
        print("This car has a " + str(self.battery_size) + "-kWh battery.")    
```

```python
ec1 = ElectricCar("Tesla", "E1", 2020, 90)
```

## 2. Multilevel Inheritance:

Class A    ---> Parent   
Class B inherit Class A    --->  Child class  
Class C inherti Class B    --->  Grandchild  

```python
# Grandchild class
class Tesla(ElectricCar):

    def __init__(self, make, model, year, battery_size):

        super().__init__(make, model, year, battery_size)

        self.autopilot = True

    def describe_autopilot(self):

        if self.autopilot:
            print("This car has Autopilot functionality.")
        else:
            print("Autopilot is not available.")
```

```python
t1 = Tesla("Honda", "HC-1", 2020, 90)
```

```python
t1.describe_autopilot()
```

### Multiple Inheritance:

```python
# Parent class 1
class Mother:
    # constructor
    def __init__(self, name, eye_color):
        # Instance attribute
        self.name = name
        self.eye_color = eye_color
        self.mother_language = "Urdu"

    # Instance Method
    def speak(self):
        return f"Mother can speak: {self.mother_language}"


# Parent class 2
class Father:
    # constructor
    def __init__(self, name):
        # Instance attributes
        self.name = name
        self.height = 6.2
        self.father_language = "Punjabi"

    # Instance Method
    def speak(self):
        return f"Father can speak: {self.father_language}"

# child class
class Child(Father, Mother):
    # constructor
    def __init__(self, mother_name, father_name, child_name, mother_eye_color):
        Mother.__init__(self, mother_name, mother_eye_color)
        Father.__init__(self, father_name)
        # Extended attributs
        self.child_name = child_name

    # Instance Method
    def additional_language(self, new_language):
        return f"Child decided to learn {new_language}"
```

```python
child_bio_data = Child("Mariam", "Raza", "Umer", "brown")

#
print(f"child inherited eye color from mother: {child_bio_data.eye_color}")
print(f"child inherited height from Father: {child_bio_data.height}")
print(f"child can speak {child_bio_data.mother_language} and {child_bio_data.father_language}")        

print(child_bio_data.speak())

print(child_bio_data.additional_language("Chinese"))
```
