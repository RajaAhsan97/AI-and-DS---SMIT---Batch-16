# Student Records Management System (SRMS)

## functionality
    1. Create Record
    2. Retrieve Record
    3. Update Record
    4. Delete Record

```python
# list of Dictionary
[
    {
        "Name": "Ahsan",
        "Rollno": "AI1",
        "marks": 90
    }, 
    {
        "Name": "Ahsan",
        "Rollno": "AI1",
        "marks": 90
    }, 

]
```

```python
student_records = []
```

```python
# add student 
def add_student():
    name = input("Enter name: ")
    rollno = input("Enter rollno: ")
    course = input("Enter course: ")
    marks = int(input("Enter marks: "))
    
    student = {
        "Name": name,
        "rollno": rollno,
        "course": course,
        "marks": marks
    }

    student_records.append(student)
```

```python
def display_record():
    if not student_records:
        print("Records are empty")
    else:
        for student in student_records:
            print("-------------------------------")
            print(f"Student Record:")
            for key, value in student.items():
                print(f"{key}: {value}")
            print("--------------------------------")
                
```

```python
# menu driven program 
while True:
    print("----- SRMS ------")
    print("1. Press {c} to add record")
    print("2. Press {r} to retrieve record")
    print("3. Press {u} to update record")
    print("4. Press {d} to delete record")
    print("5. Press {e} to exit")
    
    option = input("Enter your choice:").lower()
    
    if option == "c":
        # add student
        add_student()
    elif option == "r":
        # get student record
        display_record()
    elif option == "u":
        # update marks of student
        pass
    elif option == "d":
        # delete student record
        pass
    elif option == "e":
        break
    else:
        print("Invalid choice")

    
```
