# Student Records Management System (SRMS)

## functionality
    1. Create Record
    2. Retrieve Record
         a. Display all records
         b. Display particular student record by its rollno (Search)
    3. Update Record
    4. Delete Record

```python
# list of Dictionary
rollno = 'AI2'
[
    {
        "Name": "Ahsan",
        "Rollno": "AI1",
        "marks": 90
    }, 
    {
        "Name": "Saad",
        "Rollno": "AI2",
        "marks": 90
    }, 

]
```

```python
student_records = []
```

```python
# add student 
def add_student():
    name = input("Enter name: ")
    rollno = input("Enter rollno: ")
    course = input("Enter course: ")
    marks = int(input("Enter marks: "))
    
    student = {
        "Name": name,
        "rollno": rollno,
        "course": course,
        "marks": marks
    }

    student_records.append(student)
```

```python
# displayh all students
def display_record():
    if not student_records:
        print("Records are empty")
    else:
        for student in student_records:
            print("-------------------------------")
            print(f"Student Record:")
            for key, value in student.items():
                print(f"{key}: {value}")
            print("--------------------------------")
                
```

```python
# search student --> function definition
def search_student(roll_no):
    if not student_records:
        print("Unable to search in empty records")
    else:
        for student in student_records:
            if student["rollno"] == roll_no:
                for key, value in student.items():
                    print(f"{key}: {value}")
        
```

```python
# update student marks
def update_marks(search_rollno):
    if not student_records:
        print("Unable to upadate marks in empty records")
    else:
        for index, student in enumerate(student_records):
            if student["rollno"] == search_rollno:
                new_marks = int(input("Enter marks to update:"))

                student_records[index].update({"marks": new_marks})

                print("Marks updated successfully")
                
```

```python
# delete record
def delete_record(roll_no):
    if not student_records:
        print("Unable to delete in empty records")
    else:
        for student in student_records:
            if student["rollno"] == roll_no:
                student_records.remove(student)
                print("Remove successfully")
```

```python
# menu driven program 
while True:
    print("----- SRMS ------")
    print("1. Press {c} to add record")
    print("2. Press {r} to retrieve record")
    print("3. Press {s} to search student record")
    print("4. Press {u} to update record")
    print("5. Press {d} to delete record")
    print("6. Press {e} to exit")
    
    option = input("Enter your choice:").lower()
    
    if option == "c":
        # add student
        add_student()
    elif option == "r":
        # get student record
        display_record()
    elif option == "s":
        search_rollno = input("Enter rollno to search")
        # Function calling
        search_student(search_rollno)
    elif option == "u":
        # update marks of student
        search_rollno = input("Enter rollno to update marks")
        update_marks(search_rollno)
    elif option == "d":
        # delete student record
        search_rollno = input("Enter rollno to delete record")
        delete_record(search_rollno)
    elif option == "e":
        break
    else:
        print("Invalid choice")

    
```
