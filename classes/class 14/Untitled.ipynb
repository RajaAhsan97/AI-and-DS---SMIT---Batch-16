```python
discard_dishes = ["fish", "nihari", "qeema"]
menu = []
count = 0
while True:
    dish = input("Enter dish name: ").lower()

    if dish in discard_dishes:
        print("This is discarded dish")
        continue

    menu.append(dish)
    count += 1
    print(f"{count}: {dish} added in menu")

    if count == 5:
        break
print(menu)
```
