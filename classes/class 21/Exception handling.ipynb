## try: except:

```python
try:
    file_name = input("Enter file name: ")
    with open(file_name, "r") as file:
        print(file.read())
except FileNotFoundError:
    print("File does not exist in directory")

        
```

```python
my_list = []
print(my_list[3])
```

```python
try:
    file_name = input("Enter file name: ")
    my_list = [1,2,4,5,6]
    print(my_list[5])
    with open(file_name, "r") as file:
        print(file.read())
except FileNotFoundError:
    print("File does not exist in directory")
except IndexError:
    print("List index out of range")
```

```python
try:
    file_name = input("Enter file name: ")
    my_list = [1,2,4,5,6]
    print(my_list[5])
    with open(file_name, "r") as file:
        print(file.read())
except Exception as e:
    print(f"Exception caught: {e}")

```

```python
try: 
    num1 = int(input("Enter a number"))
    num2 = int(input("Enter a number"))

    result = num1 / num2
    print(result)
except Exception as e:
    print(f"Exception caught: {e}")
```

```python
try: 
    num1 = int(input("Enter a number"))
    num2 = int(input("Enter a number"))

    r = num1 / num2    
except Exception as e:
    print(f"Exception caught: {e}")
else:
    print(r)
```

```python
try:
    file = open("myfile.txt", "w")

    name = input("Enter name:")
    file.write(name)
    age = input("Enter age: ")
    file.write(age)
except Exception as e:
    print(f"{e}")
finally:
    file.close()
    
    
```

## User Defined Exception

```python
try:
    salary = int(input("Enter salary: "))   # 50000 < salary < 100000

    if (salary < 50000) or (salary>100000):
        raise Exception("Salary Error")
except Exception as e:
    print(e)
```
