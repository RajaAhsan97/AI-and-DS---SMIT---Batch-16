## Write a program to analyze string 
    - len
    - words
    - sentences
    - paragraphs
    - alpahabets
    - upper case
    - lower case
    - vowels
    - spaces

```python
def analyze_text(text):
    length = len(text)

    words = len(text.split(" "))

    sentences = len(text.split("."))

    paragraphs = len(text.split("\n"))

    vowels = ["a", "e", "i", "o", "u", "A", "E", "I", "O", "U"]
    alphabets = 0
    spaces = 0
    numbers = 0
    count_vowels = 0
    count_upper  = 0
    count_lower = 0
    for char in text:
        if char.isalpha():
            alphabets += 1

            if char.isupper():
                count_upper += 1
            else:
                count_lower += 1
            
            if char in vowels:
                count_vowels +=1 
        elif char.isspace():
            spaces += 1
        elif char.isdigit():
            numbers += 1

    print(f"""
    Length: {length}
    words: {words}
    sentences: {sentences}
    Vowels: {count_vowels}
    Alphabets: {alphabets}
    Upper case: {count_upper}
    Lower case: {count_lower}
    Spaces: {spaces}
    Numbers: {numbers}
    """)
```

```python
txt = """
This book would not have been possible without the wonderful and
extremely professional staff at No Starch Press. Bill Pollock invited me to
write an introductory book, and I deeply appreciate that original offer.
Tyler Ortman helped shape my thinking in the early stages of drafting.
Liz Chadwick’s and Leslie Shen’s initial feedback on each chapter was
invaluable, and Anne Marie Walker helped to clarify many parts of the
book. Riley Hoffman answered every question I had about the process of
assembling a complete book and patiently turned my work into a beautiful
finished product.
"""
```

```python
analyze_text(txt)
```
