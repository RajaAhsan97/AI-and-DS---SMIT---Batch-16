```python
# bluprint of object
class Computer():
    # varaibles/characteristics/properties/attributes
    # hard coded
    ram = "8GB"
    storage = "500GB"
    processor = "corei7"
    generation = "11th gen"
    price = 75000
    model = "HP"

    # behaviours
    def coding(self):
        print("I can do coding")

    def designing(self):
        print("I do graphics designing")
        
```

```python
c1 = Computer()
```

```python
c2 = Computer()
```

```python
c2.model
```

```python
c2.coding()
```

## constructor

```python
# bluprint of object
class Computer():
    # class attributes
    category = "IT"
    
    def __init__(self, ram, storage, processor, generation, price, model):
        # instance attribute
        # varaibles/characteristics/properties/attributes
        self.ram = ram
        self.storage = storage
        self.processor = processor
        self.generation = generation
        self.price = price
        self.model = model

    # Instance methods
    def coding(self):
        print(f"I can do coding in {self.model} {self.processor} {self.generation}")

    def designing(self):
        print(f"I do graphics designing in {self.model} {self.processor} {self.generation}")

```

```python
c1 = Computer("8gb", "500gb", "corei7", "12th gen", 80000, "Dell")
```

```python
c2 = Computer("16gb", "1Tb", "corei7", "12th gen", 100000, "hp")
```

```python
c1.category
```

```python
c2.category
```

```python
c1.coding()
```

```python
c2.coding()
```
