## What is Inheritance?

```python
# Parent class
class Car():
    def __init__(self, make, model, year):
        # attributes
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0

    # methods
    def get_descriptive_name(self):
        long_name = str(self.year) + ' ' + self.make + ' ' + self.model
        return long_name.title()
    
    def read_odometer(self):
        print("This car has " + str(self.odometer_reading) + " miles on it.")
    
    def update_odometer(self, mileage):
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can't roll back an odometer!")
```

```python
# child  class
class ElectricCar(Car):
    pass
```

```python
e1 = ElectricCar("Toyota", "e1", 2023)
```

```python
e1.get_descriptive_name()
```

```python
# child  class ---> us ky apny bi attributes hotey hein
class ElectricCar(Car):
    def __init__(self, make, model, year, battery_power, battery_type):
        super().__init__(make, model, year)
        self.battery_power = battery_power
        self.battery_type = battery_type

    def describe_battery(self):
        print(f"This car has {self.battery_power} and Type: {self.battery_type}")

    def get_descriptive_name(self):
        print(f"I am Electric car with battery power {self.battery_power}")
```

```python
ec1 = ElectricCar("Toyota", "HC1", 2019, "125KWh", "Lithium")
```
