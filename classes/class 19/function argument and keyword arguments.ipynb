```python
def add(a,b,c):
    return a+b+c
```

```python
add(2,3,5)
```

# Arbitrary number of arguments

```python
def add_num(*a):    # Receive all value in the form of tuple
    return sum(a)
```

```python
add_num(2,3)
```

```python
def add_num(*a):    # Receive all value in the form of tuple
    result = 0
    for number in a:
        if number % 2 == 0:
            result += number

    return result
```

```python
add_num(1,2,3,4,5,6,7,8)
```

# keyword arguments

```python
def my_profile(**data):
    print(type(data))
    print(data)
```

```python
my_profile(name="Ahsan", qualification="MS")
```

```python
def make_my_profile(name, subject, qualification, experience, **other_info):
    profile = {}

    profile["Name"] = name
    profile["Subject"] = subject
    profile["Qualification"] = qualification
    profile["Experience"] = experience
    profile.update(other_info)

    return profile
```

```python
make_my_profile(name="Ahsan", subject="DS", qualification="MS", experience=5, 
                city="Karachi", age=10, course="AI and DS")
```
