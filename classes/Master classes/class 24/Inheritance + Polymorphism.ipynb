## Multiple inheritance

Multiple inheritance means a child class inherits from more than one parent class.

For example, an ElectricCar can inherit features from both Car and Battery.

```python
# Parent class 1
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def get_descriptive_name(self):
        long_name = str(self.year) + ' ' + self.make + ' ' + self.model
        return long_name.title()

    def drive(self):
        print("The car is driving.")
```

```python
# Parent class 2
class Battery:
    def __init__(self, battery_size, power, manufacturer):
        self.battery_size = battery_size
        self.power = power
        self.manufacturer = manufacturer

    def describe_battery(self):
        print("This car has a " + str(self.manufacturer) + "battery, having" + str(self.battery_size) + "-kWh battery.")

    def charge(self):
        print("The battery is charging.")
```

```python
# Child class
class ElectricCar(Car, Battery):

    def __init__(self, make, model, year, battery_size, power, manufacturer):
        Car.__init__(self, make, model, year)
        Battery.__init__(self, battery_size, power, manufacturer)

    def description(self):
        print(f"{self.model} {self.make} {self.year} {self.battery_size} {self.power} {self.manufacturer}")
```

```python
ec1 = ElectricCar("Civic", "e1", 2019, 90, 180, "Exide")
```

```python
ec1.description()
```

## Multi-level Inheritance:
Multilevel inheritance means inheritance happens in multiple levels.

```python
# Parent class
class Car:

    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def get_descriptive_name(self):
        long_name = str(self.year) + ' ' + self.make + ' ' + self.model
        return long_name.title()

    def drive(self):
        print("The car is driving.")
```

```python
# Child class
class ElectricCar(Car):

    def __init__(self, make, model, year, battery_size):
        super().__init__(make, model, year)
        self.battery_size = battery_size

    def describe_battery(self):
        print("This car has a " + str(self.battery_size) + "-kWh battery.")
```

```python
# Grandchild class
class Tesla(ElectricCar):

    def __init__(self, make, model, year, battery_size, autopilot):
        super().__init__(make, model, year, battery_size)
        self.autopilot = autopilot

    def describe_autopilot(self):
        print("Autopilot: " + self.autopilot)
```

```python
ec1 = Tesla("Civic", "hc1", 2019, 90, "Yes")
```

```python
ec1.make
```

```python
ec1.drive()
```

```python
ec1.describe_battery()
```

```python
ec1.describe_autopilot()
```

Write Python program that creates a class called Employee with properties for name and salary. Include the method to calculate annual salary. Create a subclass called Manager that inherites from the Employee class and adds an additional property for department. Override the annual salary calculation method to include bonus for manager, Create two intances of the manager class and calculate their annual salary

```python
class Employee():
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary

    def calculate_annual_salary(self):
        return self.salary * 12
    
```

```python
E1 = Employee("Ahsan", 50000)
```

```python
class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary)
        self.department = department

    def calculate_annual_salary(self, bonus):
        return self.salary * 12 + bonus
```

```python
m1 = Manager("Ahsan", 50000, "IT")
```
