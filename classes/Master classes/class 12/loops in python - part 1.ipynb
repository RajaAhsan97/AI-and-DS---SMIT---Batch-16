```python
students = ["Ahsan", "Feroz", "Saqib", "Noor", "Azeem"]
```

```python
# iterating over list
for student in students:
    print(f"Hello {student}")
```

```python
for n in range(5):
    print(n)
```

```python
# iterating over list using index
for n in range(5):
    print(f"Hello {students[n]}")
```

```python
# iterating over list using index
for n in range(len(students)):
    print(f"{n} Hello {students[n]}")
```

```python
# to print list items in reverse order - we will reverse list first
for student in students[::-1]:
    print(f"Hello {student}")
```

```python
# enumerate - to get loop number
for number, student in enumerate(students[::-1]):
    print(f"{number} Hello {student}")
```

### continue , break

```python
for s in students:
    if len(s) <= 4:
        print(f"Dear Mr. {s}, kal apki chutti hy")
    else:
        print(f"Dear Mr. {s}, kal apko ana h")
```

```python
for s in students:
    if s == "Saqib":
        break
    else:
        print(s)
```

```python
# to avoid duplicate names

guests = []

for guest in range(5):
    name = input("Enter name:")

    if name in guests:
        print("Duplicate found:")
    else:
        guests.append(name)
```

```python
for a in range(10):
    if a == 5:
        continue
    else:
        print(a)
```

```python
menu = []

for a in range(5):
    dish = input("Enter dish name:").lower()

    if dish == "fish":
        continue
    menu.append(dish)
    
```

```python
menu
```

```python
rates = {}

for item in menu:
    price = int(input(f"{item} for 300 people:"))
    rates[item] = price 

rates
```

```python
# method 1 for total price
total_price = 0

for key, value in rates.items():
    total_price = total_price + value

print("Total price: ", total_price)
```

```python
# method 2 for total price
sum(rates.values())
```

```python
# combining program
discard_menu = ["fish", "qeema", "nihari"]
menu = []
mehman = 300

menu_price = {}

for i in range(5):
    # input dish name
    item = input("Enter Dish name: ").lower()

    if item in discard_menu:
        continue

    # add menu
    menu.append(item)

    # enter per plate price
    price = int(input(f"{item} for {mehman} mehman price: "))
    
    menu_price[item] = price
```

```python
menu
```

```python
menu_price
```

```python
# calculate total price
total_price = 0
for price in menu_price.values():
    total_price += price

print(total_price)
```

```python
# for per head cost
menu_price_per_head = {}

for dish, price in menu_price.items():
    menu_price_per_head[dish] = price / mehman
    
```

```python
menu_price_per_head
```

```python
print(f"Per head menu cost {sum(menu_price_per_head.values())}")
```
