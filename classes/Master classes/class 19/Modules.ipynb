```python
# menu driven program 
while True:
    print("----- SRMS ------")
    print("1. Press {c} to add record")
    print("2. Press {r} to retrieve record")
    print("3. Press {s} to search student record")
    print("4. Press {u} to update record")
    print("5. Press {d} to delete record")
    print("6. Press {e} to exit")
    
    option = input("Enter your choice:").lower()
    
    if option == "c":
        # add student
        SRMS.add_student()
    elif option == "r":
        # get student record
        SRMS.display_record()
    elif option == "s":
        search_rollno = input("Enter rollno to search")
        # Function calling
        SRMS.search_student(search_rollno)
    elif option == "u":
        # update marks of student
        search_rollno = input("Enter rollno to update marks")
        SRMS.update_marks(search_rollno)
    elif option == "d":
        # delete student record
        search_rollno = input("Enter rollno to delete record")
        SRMS.delete_record(search_rollno)
    elif option == "e":
        break
    else:
        print("Invalid choice")
```
