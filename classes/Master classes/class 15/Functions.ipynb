### Q: Write a python function that takes string as argument, and the function analyze string:
        - length of string
        - number of words
        - number of sentences
        - number of paragraphs
        - number of alphabets
        - number of upper case characters
        - number of lower case characters
        - number of vowels
        - number of spaces 
        - numbers
        - number of punctuations

```python
import string
```

```python
string.punctuation
```

```python
def text_analyzer(text):
    length = len(text)
    
    words = len(text.split(" "))
    
    sentences = len(text.split("."))
    
    paragraphs = len(text.split("\n"))
    
    vowels = ["a", "e", "i", "o", "u", "A", "E", "I", "O", "U"]
    punctuations = string.punctuation
    alphabets = 0
    spaces = 0
    numbers = 0
    lower_case = 0
    upper_case = 0
    count_vowels = 0
    count_punctuations = 0
    others = 0
    for char in text:
        if char.isalpha():
            alphabets += 1
            if char.isupper():
                upper_case += 1
            else:
                lower_case += 1

            if char in vowels:
                count_vowels += 1
                
        elif char.isspace():
            spaces += 1
        elif char.isdigit():
            numbers += 1
        elif char in punctuations:
            count_punctuations += 1
        else:
            others += 1
            print(f"{char}")

    print(f"""
    Text length: {length}
    words: {words}
    sentences: {sentences}
    paragraphs: {paragraphs}
    Alphabets: {alphabets}
    Spaces: {spaces}
    Numbers: {numbers}
    Upper case: {upper_case}
    Lowercase: {lower_case}
    Vowels: {count_vowels}
    Puncutations: {count_punctuations}
    Others: {others}
    """)
```
